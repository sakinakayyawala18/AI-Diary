```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"
```

```python
res = requests.get(URL)
```

```python
if res.status_code == 200:
    print(res.headers)

with open("scrapped_data/data1.html", "w") as f:
    f.write(res.text)
```

```python
#code 200 - ok 301/3012 - no direct, 400 - bad request, 401 - unathorized, 403 - forbidden , 404 - not found, 500 - servor error
```

```python
# BeautifulSoup

from bs4 import BeautifulSoup

with open("scrapped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    all_countries.append([name, population])
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns=["Name", "Population"])
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```
