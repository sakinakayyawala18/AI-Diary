```python
import seaborn as sns
```

```python
# sns.get_dataset_names()
sns.set_theme(style="white")
tips = sns.load_dataset('tips')
```

```python
tips.head(2)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker",
    kind="line"
)
```

```python
#relationalplot (relplot)
sns.lineplot(
    data=tips,
    x="day",
    y="tip",
    hue="smoker",
    ci=None,
    marker="o"
)
```

```python
#relationalplot (relplot)
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)
```

```python
#categorical
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex",
    edgecolor="black"
)
```

```python
#categorical
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
#matrix plot
data = sns.load_dataset('flights')
flights  = data.pivot(index="month", columns="year", values="passengers")
```

```python
import matplotlib.pyplot as plt
sns.heatmap(
    flights,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("Passenger Heatmap")
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First plot
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ci=None,
    ax=ax[0]
)

ax[0].set_title("Day vs Total Bill")

# Second plot
sns.lineplot(
    data=tips,
    x="day",
    y="tip",
    hue="smoker",
    ci=None,
    ax=ax[1]
)

ax[1].set_title("Day vs Tip")

plt.tight_layout()
plt.show()
```
