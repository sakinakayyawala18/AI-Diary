```python
#oscar_movie
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The king sppech",
    "The Artist",
    "Argo"
]

year = [2008, 2009, 2010, 2011, 2012]
oscar_revenue  = [1005, 170, 427, 133, 232]

#non oscar movies
non_oscar_movies = ["Slumdog Millionaire", "Avtar", "Inception", "Hugo", "Lincon"]

non_oscar_revenue = [378, 2788, 829, 185, 275]
```

```python
plt.plot(year, oscar_revenue, "o--r",label="oscar_movies")
plt.plot(year, non_oscar_revenue,">--g",label="non_oscar_movies")
plt.title("oscar_revenue vs non_oscar_revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.xlim(min(year)-1,max(year)+1)
plt.legend()
plt.grid()
plt.tight_layout()
plt.style.use("default")
```

```python
#Bar chart
#oscar_movie
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The king sppech",
    "The Artist",
    "Argo"
]

year = [2008, 2009, 2010, 2011, 2012]
oscar_revenue  = [1005, 170, 427, 133, 232]

#non oscar movies
non_oscar_movies = ["Slumdog Millionaire", "Avtar", "Inception", "Hugo", "Lincon"]

non_oscar_revenue = [378, 2788, 829, 185, 275]
```

```python
import numpy as np
x = np.arange(len(year))
width = 0.25
plt.style.use("default")
plt.bar(x+width/2, oscar_revenue, width, label="oscar_movie", edgecolor="black")
plt.bar(x-width/2, non_oscar_revenue,width, label="non_oscar_movie", edgecolor="black")
plt.title("oscar_revenue vs non_oscar_revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(x, year)
```

```python
plt.style.use("default")
plt.bar(year, oscar_revenue,label="Oscar Movie",edgecolor="black")
plt.title("Oscar Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
for i in range(len(year)):
    plt.text(year[i],oscar_revenue[i],str(oscar_revenue[i]),ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt

movie = ["Avatar", "Titanic", "Barbie", "Oppenheimer"]
sales = [40, 25, 20, 15]

# Explode only the Oppenheimer slice
explode = [0, 0, 0, 0.1]

plt.figure(figsize=(7,7))

plt.pie(
    sales,
    labels=movie,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True,
    explode=explode,

    wedgeprops={
        "linestyle": "--",   # Border style
        "linewidth": 2,      # Border thickness
        "edgecolor": "black",  # Border color
        "alpha": 0.8         # Transparency
    }
)

plt.title("Movie Sales Distribution", fontsize=16, fontweight="bold")

plt.show()
```

```python
# Scatter Plot
age = [18, 20, 22, 24, 26, 28]
ibm = [4.5, 5.2, 6.8, 7.1, 8.3, 9.0]
people = ["Aman", "Riya", "Rahul", "Sneha", "Ali", "Priya"]

plt.figure(figsize=(8, 6))

plt.scatter(
    age,
    ibm,
    c=ibm,
    cmap="plasma",
    s=100,
    edgecolor="black"
)

# Color bar
plt.colorbar(label="IBM")

# Title and labels
plt.title("Age vs IBM")
plt.xlabel("Age")
plt.ylabel("IBM")

# Annotate each point
for i in range(len(people)):
    plt.annotate(
        people[i],
        xy=(age[i], ibm[i]),        
        xytext=(age[i] + 0.1, ibm[i] + 0.1) 
    )

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#histogram
marks = [
    35, 42, 45, 48, 50,
    52, 55, 57, 58, 60,
    61, 63, 65, 66, 68,
    70, 72, 74, 75, 77,
    78, 80, 82, 84, 85,
    86, 88, 90, 92, 95
]

plt.hist(
    marks,
    bins=10,
    color="pink",
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y")

plt.axvline(x=70, label="above average", color="green", linestyle="--")

plt.show()
```

```python
#Boxplot
marks = [
    35, 42, 45, 48, 50,
    52, 55, 57, 58, 60,
    61, 63, 65, 66, 68,
    70, 72, 74, 75, 77,
    78, 80, 82, 84, 85,
    86, 88, 90, 92, 95
]

plt.boxplot(
    marks,
    patch_artist=True,
    widths=0.5,
    showmeans=True,
    meanline=True,
    vert=True
)

plt.title("Student Marks Box Plot")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```

```python
#suplots concept 
import numpy as np
x=[1,2,3,4,5,6]
y1=[i*i for i in x]
y2=[i*2 for i in x]
y3=[i*3 for i in x]

plt.subplot(2,2,1)
plt.plot(x,y1)
plt.title("square graph")
plt.xlabel("x")
plt.ylabel("square value")

plt.subplot(2,2,2)
plt.plot(x,y2)
plt.title("double graph")
plt.xlabel("x")
plt.ylabel("x 2 value")

plt.subplot(2,2,3)
plt.plot(x,y3)
plt.title("triple graph")
plt.xlabel("x")
plt.ylabel("x 3 value")

plt.subplot(2,2,4)
plt.plot(x,x)
plt.title("line graph")
plt.xlabel("x")
plt.ylabel("linear value")

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["Mumbai", "Delhi", "Pune", "Bangalore"]

temperature = [
    [32, 33, 31, 34, 35],   # Mumbai
    [38, 39, 40, 41, 39],   # Delhi
    [28, 29, 30, 29, 28],   # Pune
    [25, 26, 27, 26, 25]    # Bangalore
]

# Create subplots
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

k = 0

for i in range(2):
    for j in range(2):

        ax[i][j].plot(days, temperature[k], marker="o")

        ax[i][j].set_title(cities[k])
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperature (°C)")
        ax[i][j].grid(True)

        k = k + 1

plt.tight_layout()
plt.show()
```
